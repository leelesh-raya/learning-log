# Polynomial Regression — Feature Engineering

Demonstrates why linear regression fails on non-linear data
and how feature engineering fixes it without changing the algorithm.

```python
import numpy as np
import matplotlib.pyplot as plt
import copy, math
```

## Create Training Data

y = 1 + x² — a quadratic relationship.
A straight line cannot fit this data.

```python
x_train = np.arange(0,20,1)
y_train = 1 + x_train**2
```

## Cost Function

Vectorised — no loops needed since x and y are 1D arrays.
NumPy broadcasting handles all m examples simultaneously.

```python
def compute_cost(x, y, w, b):
    cost = 0
    f_wb = w*x + b 
    cost = (f_wb - y)**2
    total_cost = cost.mean()/2
    return total_cost
```

## Gradient Function

Vectorised gradients using array operations.
.mean() divides by m, replacing the explicit loop and division.

```python
def compute_gradient(x, y, w, b):
    f_wb = w*x + b
    grad1 = (f_wb - y)*x
    dj_dw = grad1.mean()
    grad2 = (f_wb - y)
    dj_db = grad2.mean()
    return dj_dw, dj_db
```

## Gradient Descent

Iteratively updates w and b using computed gradients.
Stores cost history at each iteration for convergence tracking.

```python
def gradient_descent(x, y, w_in, b_in, alpha, num_iter, cost_func, gradient_func):
    w = copy.deepcopy(w_in)
    b = b_in
    j_history = []
    for i in range(num_iter):
        dj_dw, dj_db = gradient_func(x, y, w, b)
        w = w - alpha*dj_dw
        b = b - alpha*dj_db

        if i<(1000000):
            j_history.append(cost_func(x,y,w,b))

        if i% math.ceil(num_iter / 10)==0:
            print(f"Iteration: {i:<8} cost:{j_history[-1]:12.6f}")

    return w,b, j_history
```

## Train Without Feature Engineering

Raw x fed directly to the model.
A linear model cannot capture a quadratic relationship —
the best it can do is a straight line through the curve.

```python
initial_w = 0
initial_b = 0
alpha = 1e-4
iterations = 1000
w_final, b_final, cost = gradient_descent(x_train, y_train, initial_w, initial_b, alpha, iterations, compute_cost, compute_gradient)
```

## Plot — No Feature Engineering

The line fits poorly — it misses the curvature entirely.
This confirms the model needs a better input representation.

```python
yp = w_final*x_train + b_final
plt.scatter(x_train, y_train, marker='x', c='r')
plt.plot(x_train, yp, c='blue')
plt.title('No feature engineering')
plt.xlabel('x_train'); plt.ylabel('y')
```

## Train With Feature Engineering

Replace x with x² as the input feature.
Now the model sees a linear relationship between x² and y,
because y = 1 + x² is linear in x².
Gradient descent finds the correct w≈1 and b≈1.

```python
X = x_train**2
initial_w = 0
initial_b = 0
alpha = 1e-5
iterations = 1000
w_final, b_final, cost = gradient_descent(X, y_train, initial_w, initial_b, alpha, iterations, compute_cost, compute_gradient)
```

## Plot — With Feature Engineering

The model now fits the data well.
x_train (not X=x²) is used on the x-axis —
the plot shows predictions in the original input space.
yf was computed using X=x², so the curve appears quadratic
when viewed against the original x.

```python
yf = w_final*X + b_final
plt.scatter(x_train, y_train, marker='x', c='r')
plt.plot(x_train, yf, c='blue')
plt.title('With feature engineering')
plt.xlabel('x_train'); plt.ylabel('y')
print(b_final)
```
