## Learn to implement the $model f_{w,b}$ for linear regression with one variable


```python
import numpy as np
import matplotlib.pyplot as plt

x_train = np.array([1, 2])
y_train = np.array([300, 500])
```

m is the number of training examples

```python
m = x_train.shape[0]
print(m)
```

```python
plt.scatter(x_train, y_train, marker = 'x', c= 'r' )
plt.title("Housing Prices")
plt.xlabel("Size (1000 sqft)")
plt.ylabel('Price (in 1000s of dollars)')
plt.show()
```

Try experimenting with different values of  𝑤
  and  𝑏
 . What should the values be for a line that fits our data

```python
w = 200
b = 100
```

Let's define the function for predicting the price

```python
def formula(x , w, b):
    f_wb = np.zeros(m)
    for i in range(m):
        f_wb[i] = w * x[i] + b
    
    return f_wb
```

Now let's call the `formula` function and plot the output..

```python
f = formula(x_train , w, b)

plt.plot(x_train, f, c='r')
plt.scatter(x_train, y_train, marker= 'x', c = 'b')
plt.title("Housing Prices")
plt.ylabel('Price (in 1000s of dollars)')
plt.xlabel('Size (1000 sqft)')

plt.show()
```

### Prediction
Now that we have a model, we can use it to make our original prediction. Let's predict the price of a house with 1200 sqft. Since the units of $x$ are in 1000's of sqft, $x$ is 1.2.

```python
x_i = 1.2
cost_1200sqft = w * x_i + b
print(cost_1200sqft)
```
