# Neural Networks with Keras — Single Neuron

Demonstrating that a single neuron with sigmoid activation
is mathematically equivalent to logistic regression.
Keras is used here to understand the framework before
building multi-layer networks.

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from keras.layers import Dense, Input
from keras import Sequential
from keras.losses import MeanSquaredError, BinaryCrossentropy
from keras.activations import sigmoid
```

## Build a Single Neuron Model

Sequential model with one Dense layer and sigmoid activation.
One input feature, one output neuron.
This is the simplest possible neural network.

```python
# Sequential is used here because our network is a simple, linear pipeline.
# Data passes strictly forward from the input directly to a single layer,
# with no branching, no multiple inputs, and no complex shortcuts.
model = Sequential(
     # Defines the input structure (1 feature column, flexible batch size)
    [Input(shape=(1,)),
     # Processes the input linearly through a single neuron using Sigmoid activation
     Dense(1, activation=sigmoid, name='L1'),
     ]
    )
```

## Inspect Model Architecture

summary() shows layer names, output shapes, and parameter count.
One neuron with one input has 2 parameters — w and b.

```python
model.summary()
```

## Inspect Initial Weights

Keras initialises weights randomly by default.
get_weights() returns [w_matrix, b_array].

```python
logistic_layer = model.get_layer('L1')
w,b = logistic_layer.get_weights()
print(w, b)
print(w.shape,b.shape)
```

## Set Weights Manually

Override random initialisation with specific values.
w=2, b=-4.5 gives a decision boundary at x=2.25.
This matches the logistic regression setup from Course 1.

```python
set_w = np.array([[2]])
set_b = np.array([-4.5])
logistic_layer.set_weights([set_w, set_b])
print(logistic_layer.get_weights())
```

## Create Input Data

Six evenly spaced values from 0 to 5.
Reshaped to (6,1) — Keras expects (batch_size, n_features).

```python
X_train = np.array([0., 1, 2, 3, 4, 5], dtype=np.float32).reshape(-1,1) 
```

## Predict Using Keras

model.predict() runs the full forward pass through the network.
For a single neuron this is: sigmoid(w·x + b)

```python
a1 = model.predict(X_train[0].reshape(-1,1))
print(a1)
```

## Verify Against Manual Calculation

Manually compute sigmoid(w·x + b) using the same weights.
Result must match model.predict() exactly —
confirming the neuron is doing logistic regression.

```python
alog = sigmoid(np.dot(set_w,X_train[0].reshape(-1,1)) + set_b)
print(alog)
```
